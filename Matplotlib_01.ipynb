### Bar chart

image.png

- Bivariate Analysis
- Numerical vs Categorical
- Use case - Aggregate analysis of groups

```python
# simple bar chart
children = [10,20,40,10,30]
colors = ['red','blue','green','yellow','pink']

plt.bar(colors,children,color='green')
```

```python
plt.barh(colors,children)
```

```python
people = ['Rahim', 'Karim', 'Shium']

season_1 = [80, 50, 170]
season_2 = [85, 55, 195]
season_3 = [90, 40, 138]

x = np.arange(len(people))
width = 0.25

plt.bar(x - width, season_1, width, label='Season 1')
plt.bar(x, season_2, width, label='Season 2')
plt.bar(x + width, season_3, width, label='Season 3')

plt.xticks(x, people)
plt.xlabel('People')
plt.ylabel('Score')
plt.title('Season-wise Scores')
plt.legend()

plt.show()
print(x)
```

```python
plt.bar(x,season_1)
```

```python
plt.bar(x-.5,season_1)
```

```python
plt.bar(x+15,season_1,width=.25)
```

```python
# color and label
df = pd.read_csv('/content/batsman_season_record.csv')
df
```

```python
plt.bar(np.arange(df.shape[0])-.3,df['2015'],width=.3)
plt.bar(np.arange(df.shape[0]),df['2016'],width=.3)
plt.bar(np.arange(df.shape[0])+.3,df['2017'],width=.3)

plt.xticks(np.arange(df.shape[0]),df['batsman'])
plt.show()
```

```python
plt.bar(np.arange(df.shape[0])-.3,df['2015'],width=.3)
plt.bar(np.arange(df.shape[0]),df['2016'],width=.3)
plt.bar(np.arange(df.shape[0])+.3,df['2017'],width=.3)

plt.xticks(np.arange(df.shape[0]),df['batsman'],rotation=70)
plt.show()
```

```python
plt.bar(np.arange(df.shape[0])-.3,df['2015'],width=.3)
plt.bar(np.arange(df.shape[0]),df['2016'],width=.3)
plt.bar(np.arange(df.shape[0])+.3,df['2017'],width=.3)

plt.xticks(np.arange(df.shape[0]),df['batsman'],rotation=90)
plt.show()
```

```python
# Stacked Bar chart

plt.bar(df['batsman'],df['2017'],label='2017')
plt.bar(df['batsman'],df['2016'],bottom=df['2017'],label='2016')
plt.bar(df['batsman'],df['2015'],bottom=(df['2016'] + df['2017']),label='2015')

plt.legend()
plt.show()
```

### Pie Chart

image.png

- Univariate/Bivariate Analysis
- Categorical vs numerical
- Use case - To find contibution on a standard scale

```python
# simple data
data = [23,45,100,20,49]
subjects = ['eng','science','maths','sst','hindi']
plt.pie(data,labels=subjects)

plt.show()
```
