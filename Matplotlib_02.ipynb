### 3D Scatter Plots

```python
fig, ax = plt.subplots(nrows=2,ncols=2,figsize=(15,10),subplot_kw={'projection': '3d'})

ax[0,0].scatter3D(batters['runs'],batters['avg'],batters['strike_rate'],marker='+')
ax[0,0].set_title('IPL batsman analysis')
```

```python
fig,ax=plt.subplots(nrows=1,ncols=2,figsize=(10,7),subplot_kw={'projection': '3d'})

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

X, Y = np.meshgrid(x, y)

Z1 = X**2 + Y**2
Z2 = np.sin(np.sqrt(X**2 + Y**2))

ax[0].plot_surface(X, Y, Z1, cmap='viridis')
ax[0].set_title('3D Paraboloid')

ax[1].plot_surface(X, Y, Z2, cmap='plasma')
ax[1].set_title('3D Sine Surface')

fig.suptitle('Surface Plot')  # Overall title
plt.tight_layout()
fig.show()
```

```python
fig, ax = plt.subplots(
    nrows=1,
    ncols=2,
    figsize=(10, 7),
    subplot_kw={'projection': '3d'}
)

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

X, Y = np.meshgrid(x, y)

Z1 = X**2 + Y**2
Z2 = np.sin(np.sqrt(X**2 + Y**2))

surf1 = ax[0].plot_surface(X, Y, Z1, cmap='viridis')
ax[0].set_title('3D Paraboloid')

surf2 = ax[1].plot_surface(X, Y, Z2, cmap='plasma')
ax[1].set_title('3D Sine Surface')

fig.colorbar(surf1, ax=ax[0])
fig.colorbar(surf2, ax=ax[1], shrink=0.6)

fig.suptitle('Surface Plot')

plt.tight_layout()
plt.show()
```

### 3D Surface Plots

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
```

```python
xx, yy = np.meshgrid(x,y)
```

```python
z = xx**2 + yy**2
z.shape
```

```python
fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

### Contour Plots

```python
fig = plt.figure(figsize=(12,8))

ax = plt.subplot()

p = ax.contour(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
fig = plt.figure(figsize=(12,8))

ax = plt.subplot()

p = ax.contourf(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize=(12,8))

ax = plt.subplot()

p = ax.contourf(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```
